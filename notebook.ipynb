```python
from reportanalyser import extract_values, extract_text

data = extract_values(extract_text("report.jpg"))

print(data)
```

```python
print("\n" + "="*40)
print("🧠 HEALTH REPORT SUMMARY")
print("="*40)

if data['hemoglobin'] < 12:
    print("⚠️ Hemoglobin is LOW (", data['hemoglobin'], ") → Risk of anemia")

if data['vitamin_d'] < 20:
    print("⚠️ Vitamin D is LOW (", data['vitamin_d'], ") → Bone weakness risk")

if data['b12'] < 300:
    print("⚠️ Vitamin B12 is LOW (", data['b12'], ") → Fatigue / nerve issues")

if data['cholesterol'] > 200:
    print("⚠️ Cholesterol is HIGH (", data['cholesterol'], ") → Heart risk")

print("="*40)
```
